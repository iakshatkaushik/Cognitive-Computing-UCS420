```python
roll_no = input("Enter your roll number: ")

digits = []
for i in roll_no:
    digits.append(int(i))

L = []
for i in digits:
    L.append(i * 10)

print(L)

L.append(50)
print(L)  # append() adds an element at the end

L.insert(2, 70)
print(L)  # insert() adds an element at the given position

L.remove(50)
print(L)

L.pop(2)
print(L)

L.sort()
print(L)

L.sort(reverse=True)
print(L)

print(L[:3])
print(L[-3:])

average = sum(L) / len(L)
new_list = [i for i in L if i > average]
print(new_list)
```

```python
scores = tuple(L[:8])

highest = max(scores)
print(highest)
print(scores.index(highest))

lowest = min(scores)
print(lowest)
print(scores.count(lowest))

reverse_scores = list(scores[::-1])
print(reverse_scores)  # Tuples are immutable, so they cannot be reversed in place

score = int(input("Enter a score: "))

if score in scores:
    print(scores.index(score))
else:
    print("Score not present")

try:
    scores[0] = 100
except TypeError as e:
    print(e)  # Tuples cannot be changed after creation, unlike lists

first, second, *remaining = scores
print(first)
print(second)
print(remaining)
```

```python
import random

random.seed(int(roll_no))

numbers = []

for i in range(100):
    numbers.append(random.randint(100, 900))

odd = []
for i in numbers:
    if i % 2 != 0:
        odd.append(i)

print(len(odd))
print(odd)

even = []
for i in numbers:
    if i % 2 == 0:
        even.append(i)

print(len(even))
print(even)

prime = []

for n in numbers:
    count = 0

    for i in range(1, n + 1):
        if n % i == 0:
            count = count + 1

    if count == 2:
        prime.append(n)

prime_list = [i for i in numbers if i in prime]

print(len(prime_list))
print(prime_list)

most = numbers[0]
most_count = numbers.count(most)

for i in numbers:
    count = numbers.count(i)

    if count > most_count:
        most = i
        most_count = count

print(most)
print(most_count)
```

```python
A = set()

for i in digits:
    A.add(i * 7)

B = set()

for i in digits:
    B.add(i * 9)

print(A)
print(B)

print(A.union(B))
print(A.intersection(B))

print(A.difference(B))
print(B.difference(A))  # difference() gives elements in one set but not the other

print(A.symmetric_difference(B))

print(A.issubset(B))
print(B.issuperset(A))

X = int(input("Enter X: "))

A.discard(X)
print(A)  # discard() does not give an error if X is not present
```

```python
name = input("Enter your name: ")
branch = input("Enter your branch: ")
age = int(input("Enter your age: "))
city = input("Enter your home city: ")
cgpa = float(input("Enter your CGPA: "))

my_dict = {
    "name": name,
    "roll_no": roll_no,
    "branch": branch,
    "age": age,
    "city": city
}

my_dict["location"] = my_dict.pop("city")

my_dict["cgpa"] = cgpa

my_dict["age"] = my_dict["age"] + 1

dict1 = my_dict.copy()
dict1.pop("branch")
print(dict1)

dict2 = my_dict.copy()
del dict2["branch"]
print(dict2)

# pop() returns the deleted value, while del only deletes the key

for key, value in my_dict.items():
    print(key, "→", value)

if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email not found")

friend_dict = {
    "name": "Rahul",
    "roll_no": "1024170128",
    "branch": "CSE",
    "age": 20,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}
print(merged_dict)  # Values from friend_dict win when both dictionaries have the same key

string_dict = {key: value for key, value in my_dict.items() if type(value) == str}
print(string_dict)
```
