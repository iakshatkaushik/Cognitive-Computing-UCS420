```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
```

```python
values,count = np.unique(x,return_counts = True)
max_count = np.max(count)
most_frequent = values[count == max_count]
print("Array x:", x)
print("Most frequent value:", most_frequent)
print("Frequency:", max_count)
for value in most_frequent:
    print("Indices of", value, ":", np.where(x == value)[0])
```

```python
y= np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])
values,counts = np.unique(y,return_counts = True)
max_count = np.max(count)
most_freq = values[counts == max_count]
print("Array x:", y)
print("Most frequent value:", most_frequent)
print("Frequency:", max_count)
for value in most_frequent:
    print("Indices of", value, ":", np.where(y == value)[0])
```

```python
Akshat = np.linspace(10, 100, 25)
print("Array:")
print(Akshat)

print("\nDimensions:", Akshat.ndim)

print("Shape:", Akshat.shape)

print("Total elements:", Akshat.size)

print("Data type:", Akshat.dtype)

print("Total bytes:", Akshat.nbytes)

transpose_Akshat = Akshat.reshape(25, 1)

print("\nTranspose using reshape():")
print(transpose_Akshat)

transpose_T = Akshat.T

print("\nUsing T attribute:")
print(transpose_T)

print("\nShape using reshape:", transpose_Akshat.shape)
print("Shape using T:", transpose_T.shape)
```

```python
ucs420_Akshat = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_Akshat)

print("\nMean:", np.mean(ucs420_Akshat))

print("Median:", np.median(ucs420_Akshat))

print("Maximum:", np.max(ucs420_Akshat))

print("Minimum:", np.min(ucs420_Akshat))

print("Unique elements:", np.unique(ucs420_Akshat))

reshaped_ucs420_Akshat = ucs420_Akshat.reshape(4, 3)

print("\nReshaped Array (4 x 3):")
print(reshaped_ucs420_Akshat)

resized_ucs420_Akshat = np.resize(ucs420_Akshat, (2, 3))

print("\nResized Array (2 x 3):")
print(resized_ucs420_Akshat)
```
