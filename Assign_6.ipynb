```python
import numpy as np
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
actual_temp = temperature+2
print(f"Corrected temperature : {actual_temp}")
fahrenheit = (9/5) * actual_temp + 32
print(f"Temperature in fahrenheit : {fahrenheit}")
temp_greater_32 = np.sum(actual_temp>32)
print(f"Temperature greater than 32 : {temp_greater_32}")
```

### **Explain how vectorization and Boolean indexing can make data processing more efficient than explicitly iterating through every value.**


 **Vectorization performs operations on the entire NumPy array without explicitly using a for loop .**
 **Boolean indexing efficiently filters elements based on a condition .**

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

total_steps = steps.sum()
print(f"Total steps : {total_steps}")
mean_steps = steps.mean()
print(f"Mean of the steps : {mean_steps}")
maximum = steps.max()
minimum = steps.min()
print(f"Maximum steps is {maximum} and minimum is {minimum}")

daily_total = steps.sum(axis=0)
print(f"Total steps for each day : {daily_total}")

user_total = steps.sum(axis=1)
print(f"Total steps for each user : {user_total}")
max_index = steps.argmax()
position = np.unravel_index(max_index, steps.shape)
print("User/Day position:")
print(position)

print("Maximum steps:", steps[position])
```

```python
original = np.array([1, 2, 3, 4, 5, 6])

print("a. Original array:")
print(original)
subset = original[1:5]
print("\nb. Subset:")
print(subset)
subset[0] = 999
copied_array = original[1:5].copy()
copied_array[0] = 500
print("\nd. After modifying copied array:")
print("Original:", original)
print("Copied array:", copied_array)

matrix = np.arange(1, 13).reshape(3, 4)
print(" 3 x 4 Matrix:")
print(matrix)

first_row = matrix[0]
last_row = matrix[-1]
second_column = matrix[:, 1]
selected_elements = matrix[0:2, 1:3]

print("First row:")
print(first_row)

print("Last row:")
print(last_row)

print("Second column:")
print(second_column)

print("Rows 1-2 and Columns 2-3:")
print(selected_elements)

flatten_array = matrix.flatten()
ravel_array = matrix.ravel()

print(" Flatten:")
print(flatten_array)

print("Ravel:")
print(ravel_array)

ravel_array[0] = 100

print("\nh. After modifying ravel array:")
print("Ravel array:")
print(ravel_array)

print("Original matrix:")
print(matrix)

flatten_array[1] = 500

print("\ni. After modifying flatten array:")
print("Flatten array:")
print(flatten_array)

print("Original matrix:")
print(matrix)

print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
y = np.array([40, 65, 30, 85])
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
print("a. Shape of X:")
print(X.shape)

print("Dimensions of X:")
print(X.ndim)
XT = X.T

print(" Transpose X.T:")
print(XT)
XtX = X.T @ X

print("X.T @ X:")
print(XtX)

inverse = np.linalg.inv(XtX)

print("Inverse of X.T @ X:")
print(inverse)
beta = np.linalg.inv(X.T @ X) @ X.T @ y

print("\ne. OLS Coefficients:")
print(beta)

new_user = np.array([5, 40, 7])

predicted_score = new_user @ beta

print("\ng. New user:")
print(new_user)

print("Predicted assistance score:")
print(predicted_score)
```
