```python
iris.columns = ['id', 'sepal_len', 'sepal_wid', 'petal_len', 'petal_wid', 'species']
iris = iris.set_index('id')
iris.head()
```

```python
iris['petal_area'] = iris['petal_len'] * iris['petal_wid']
iris[['petal_len', 'petal_wid', 'petal_area']].head()
```

```python
petal_len = iris['petal_len'].values
print("Type:", type(petal_len))
print("Number of dimensions:", petal_len.ndim)
print("Size:", petal_len.size)
```

```python
print("Standard Deviation:", round(np.std(petal_len), 4))
print("Variance:", round(np.var(petal_len), 4))
print("Range:", np.ptp(petal_len))
print("75th Percentile:", np.percentile(petal_len, 75))
```

```python
petal_len_mm = petal_len * 10
print("First 10 values in mm:", petal_len_mm[:10])
```

```python
long_petals = petal_len[petal_len > 4]
print("Flowers with petal length > 4 cm:", len(long_petals))
print("Index of longest petal:", np.argmax(petal_len))
```

```python
arr = iris[['petal_len', 'petal_wid']].head(12).values
print("Original shape:", arr.shape)
print("Transposed shape:", np.transpose(arr).shape)
```

```python
new_arr = arr.reshape(4, 6)
print("Reshaped to (4, 6):\n", new_arr)
print("\nFlattened:\n", new_arr.flatten())
```
